```python
import csv
import glob
from collections import defaultdict

import cv2

imgs = sorted(glob.glob("/content/datos/seq_0004/images/*.jpeg"))
nombres = [p.split("/")[-1] for p in imgs]
alto = cv2.imread(imgs[0]).shape[0]
y_linea = int(0.5 * alto)

reg = list(csv.DictReader(open("resultados/seq_0004/registro.csv")))
pistas = defaultdict(list)
for r in reg:
    i = int(r["cuadro"])
    pistas[int(r["id"])].append((nombres[i], float(r["cy"])))

print(f"Altura imagen: {alto}, línea en y={y_linea}\n")
for tid in sorted(pistas):
    p = pistas[tid]
    lados = ["arriba" if cy < y_linea else "abajo" for _, cy in p]
    cambios = sum(1 for a, b in zip(lados, lados[1:]) if a != b)
    print(f"ID {tid}: {len(p)} detecciones, de {p[0][0]} a {p[-1][0]}, "
          f"primer lado: {lados[0]}, cambios de lado: {cambios}")
```

```python
import glob

from ultralytics import YOLO

imgs = sorted(glob.glob("/content/datos/seq_0004/images/*.jpeg"))[:12]
modelo = YOLO("weights/exp_01.pt")

for ruta in imgs:
    r = modelo.predict(ruta, conf=0.05, imgsz=640, verbose=False)[0]
    if len(r.boxes) == 0:
        print(ruta.split("/")[-1], "sin detecciones")
        continue
    detalle = [(round(float(c), 2), round(float(y), 1))
               for c, y in zip(r.boxes.conf, r.boxes.xywh[:, 1])]
    print(ruta.split("/")[-1], detalle)
```

```python
import glob

from ultralytics import YOLO

imgs = sorted(glob.glob("/content/datos/seq_0004/images/*.jpeg"))[:12]
modelo = YOLO("weights/exp_01.pt")

resultados = modelo.track(
    source=imgs,
    stream=True,
    persist=True,
    tracker="bytetrack.yaml",
    conf=0.05,
    imgsz=640,
    verbose=False,
)

for ruta, r in zip(imgs, resultados):
    nombre = ruta.split("/")[-1]
    if r.boxes is None or len(r.boxes) == 0:
        print(nombre, "sin detecciones")
        continue
    ids = r.boxes.id.int().tolist() if r.boxes.id is not None else [None] * len(r.boxes)
    filas = [
        f"ID={tid} conf={float(c):.2f} cy={float(xywh[1]):.0f}"
        for tid, c, xywh in zip(ids, r.boxes.conf, r.boxes.xywh)
    ]
    print(nombre, " | ".join(filas))
```
